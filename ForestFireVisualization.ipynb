```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# =====================================================
# LOAD DATASET
# =====================================================

forest_df = pd.read_csv("forestfires.csv")

print("First 5 Rows:")
print(forest_df.head())
```

```python
# =====================================================
# 1. HISTOGRAM
# Temperature Distribution
# =====================================================

plt.figure(figsize=(8,5))

plt.hist(forest_df['temp'], bins=20)

plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")

plt.show()
```

```python
# =====================================================
# 2. BAR PLOT
# Monthwise Burned Area
# =====================================================

plt.figure(figsize=(10,5))

sns.barplot(x='month',
            y='area',
            data=forest_df)

plt.title("Monthwise Burned Area")
plt.xlabel("Month")
plt.ylabel("Burned Area")

plt.show()
```

```python
# =====================================================
# 3. SCATTER PLOT
# Temperature vs Burned Area
# =====================================================

plt.figure(figsize=(8,5))

plt.scatter(forest_df['temp'],
            forest_df['area'])

plt.title("Temperature vs Burned Area")
plt.xlabel("Temperature")
plt.ylabel("Burned Area")

plt.show()
```

```python
# =====================================================
# 4. LINE PLOT
# Temperature Variation
# =====================================================

plt.figure(figsize=(10,5))

plt.plot(forest_df['temp'])

plt.title("Temperature Variation")
plt.xlabel("Index")
plt.ylabel("Temperature")

plt.show()
```

```python
# =====================================================
# 5. BOX PLOT
# Wind Speed Distribution
# =====================================================

plt.figure(figsize=(6,5))

sns.boxplot(y=forest_df['wind'])

plt.title("Wind Speed Distribution")

plt.show()
```

```python
# =====================================================
# 6. HEATMAP
# Correlation Matrix
# =====================================================

plt.figure(figsize=(10,8))

sns.heatmap(
    forest_df.corr(numeric_only=True),
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()
```
