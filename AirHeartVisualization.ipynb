```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Datasets
air_df = pd.read_csv("air_quality.csv")
heart_df = pd.read_csv("heart.csv")
```

```python
# =====================================================
# 1. AIR QUALITY DATASET VISUALIZATION
# =====================================================

# Histogram using matplotlib
plt.figure(figsize=(8,5))
plt.hist(air_df.iloc[:, 0], bins=20)

plt.title("Air Quality Data Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
# =====================================================
# 2. HEART DISEASE DATASET VISUALIZATION
# =====================================================

# Countplot using seaborn
plt.figure(figsize=(6,5))

sns.countplot(x='target', data=heart_df)

plt.title("Heart Disease Count")
plt.xlabel("Target")
plt.ylabel("Count")

plt.show()
```

```python
# =====================================================
# 3. HEATMAP
# =====================================================

plt.figure(figsize=(10,8))

sns.heatmap(heart_df.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# =====================================================
# 4. SCATTER PLOT
# =====================================================

plt.figure(figsize=(7,5))

plt.scatter(heart_df['age'], heart_df['chol'])

plt.title("Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")

plt.show()
```
