```python
# ============================================================
# ECOMMERCE REVIEW SCRAPPER USING PYTHON
# ============================================================

# ------------------------------------------------------------
# IMPORT LIBRARIES
# ------------------------------------------------------------

import requests
from bs4 import BeautifulSoup
import pandas as pd

print("Libraries Imported Successfully")

# ------------------------------------------------------------
# WEBSITE URL
# ------------------------------------------------------------

url = "https://quotes.toscrape.com/"

# ------------------------------------------------------------
# SEND REQUEST
# ------------------------------------------------------------

response = requests.get(url)

print("\nWebsite Status :", response.status_code)

# ------------------------------------------------------------
# PARSE HTML
# ------------------------------------------------------------

soup = BeautifulSoup(response.text, "html.parser")

# ------------------------------------------------------------
# FIND ALL REVIEW BLOCKS
# ------------------------------------------------------------

reviews = soup.find_all("div", class_="quote")

print("Total Reviews Found :", len(reviews))

# ------------------------------------------------------------
# EMPTY LISTS
# ------------------------------------------------------------

customer_names = []
ratings = []
comment_tags = []
review_comments = []

# ------------------------------------------------------------
# EXTRACT DATA
# ------------------------------------------------------------

for review in reviews:

    # --------------------------------------------------------
    # CUSTOMER NAME
    # --------------------------------------------------------

    customer = review.find("small", class_="author")

    if customer:
        customer_names.append(customer.text.strip())
    else:
        customer_names.append("N/A")

    # --------------------------------------------------------
    # REVIEW COMMENT
    # --------------------------------------------------------

    comment = review.find("span", class_="text")

    if comment:
        review_comments.append(comment.text.strip())
    else:
        review_comments.append("N/A")

    # --------------------------------------------------------
    # COMMENT TAGS
    # --------------------------------------------------------

    tags = review.find_all("a", class_="tag")

    tag_list = []

    for tag in tags:
        tag_list.append(tag.text)

    comment_tags.append(", ".join(tag_list))

    # --------------------------------------------------------
    # RANDOM RATINGS (FOR DEMO PURPOSE)
    # --------------------------------------------------------

    import random

    ratings.append(random.randint(1, 5))

# ------------------------------------------------------------
# CREATE DATAFRAME
# ------------------------------------------------------------

df = pd.DataFrame({
    "Customer Name": customer_names,
    "Rating": ratings,
    "Comment Tag": comment_tags,
    "Review Comment": review_comments
})

# ------------------------------------------------------------
# DISPLAY FINAL DATA
# ------------------------------------------------------------

print("\n========== FINAL REVIEW DATA ==========\n")

print(df)

# ------------------------------------------------------------
# SAVE CSV FILE
# ------------------------------------------------------------

df.to_csv("review_data.csv", index=False)

print("\nCSV File Saved Successfully")

# ------------------------------------------------------------
# SAVE EXCEL FILE
# ------------------------------------------------------------

df.to_excel("review_data.xlsx", index=False)

print("Excel File Saved Successfully")

# ------------------------------------------------------------
# TOTAL REVIEWS
# ------------------------------------------------------------

print("\nTotal Reviews Scrapped :", len(df))

# ------------------------------------------------------------
# RATING COUNTS
# ------------------------------------------------------------

print("\n========== RATING COUNTS ==========\n")

print(df["Rating"].value_counts())

# ============================================================
# END OF PROJECT
# ============================================================
```
